```python
# Cell 1: imports
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Cell 2: load and clean
df = pd.read_csv("titanic.csv")
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
df = df.drop_duplicates()

# Cell 4: chart 2, age distribution (independent of 'Survived')
plt.hist(df["Age"], bins=20, edgecolor="black")
plt.title("Age distribution")
plt.xlabel("Age")
plt.show()

# Conditional plotting for survival-related charts
if 'Survived' not in df.columns:
    print("Warning: 'Survived' column not found in the DataFrame. Survival-related plots cannot be generated.")
else:
    # Cell 3: chart 1, survival by sex
    df.groupby("Sex")["Survived"].mean().plot(kind="bar", title="Survival rate by sex")
    plt.ylabel("Rate")
    plt.show()

    # Cell 5: chart 3, survival by class
    if 'Pclass' not in df.columns:
        print("Warning: 'Pclass' column not found in the DataFrame. Cannot generate survival rate by class plot.")
    else:
        df.groupby("Pclass")["Survived"].mean().plot(kind="bar", color="green")
        plt.title("Survival rate by class")
        plt.show()

    # Cell 6: chart 4, fare vs age
    plt.scatter(df["Age"], df["Fare"], c=df["Survived"], alpha=0.5)
    plt.xlabel("Age")
    plt.ylabel("Fare")
    plt.title("Fare vs age (colored by survival)")
    plt.show()
```
